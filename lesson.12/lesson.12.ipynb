```python
temp_list = [
    18.2, 18.9, 19.4,
    18.0, 18.7, 19.1,
    17.7, 18.6, 19.0,
    18.1, 18.9, 19.4,
    18.2, 18.2, 19.8,
    13.4, 18.9, 16.4,
    19.2, 14.1, 15.4,
    12.6, 16.9, 14.9,
    17.8, 13.1, 19.6,
    14.4, 18.2, 18.7,
    12.2, 12.6, 14.2,
    13.5, 17.1, 16.3,
    14.1, 12.5, 12.4,
    18.2, 14.7, 11.4,

]
print(temp_list)
```

```python
temp_values = np.array(temp_list)
print(temp_values)
print(temp_values.ndim)
print(temp_values.shape)
print(temp_values.size)
```

```python
temp_days = temp_values.reshape(14, 3)
print(temp_days)
print(temp_days.ndim)
print(temp_days.shape)
print(temp_days.size)
```

```python
temp_days_1 = temp_values.reshape(-1, 3)
print(temp_days_1)
print(temp_days_1.ndim)
print(temp_days_1.shape)
print(temp_days_1.size)
```

```python
temp_days_3d = temp_values.reshape(2, 7, 3)
print(temp_days_3d)
print(temp_days_3d.ndim)
print(temp_days_3d.shape)
print(temp_days_3d.size)
```

```python
#resize

base_array = np.array([10, 20, 30, 40, 50, 60])
print(base_array)
```

```python
resh_array = base_array.reshape(2, 3)
print(resh_array)
print(resh_array.ndim)
print(resh_array.shape)
print(resh_array.size)
```

```python
resize_array =  np.resize(base_array, 7)
print(resize_array)
print(resize_array.ndim)
print(resize_array.shape)
print(resize_array.size)
```

```python
resh_array = resize_array.reshape(4, 3)
print(resh_array)
print(resh_array.ndim)
print(resh_array.shape)
print(resh_array.size)
```
