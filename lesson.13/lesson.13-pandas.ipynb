```python
# Маски
my_mask = df["Price_euros"] < 1000
my_mask.head()
```

```python
df.head()
```

```python
type(my_mask)
```

```python
cheap_laptops = df[my_mask]
cheap_laptops.head(10)
```

```python
apple_mask = df["Company"] == "Apple"
cheap_apple = df[apple_mask & my_mask]
cheap_apple.head(10)
```

```python
apple_mask = df["Company"] == "Apple"
cheap_apple = df[apple_mask | my_mask]
cheap_apple[["Company", "laptop_ID", "Price_euros" ]].head(10)
```

```python
apple_mask = df["Company"] == "Apple"
cheap_apple = df[~ apple_mask ]
cheap_apple[["Company", "laptop_ID", "Price_euros" ]].head(10)
```

```python
def get_price_group(price):
    if price < 1000:
        return "До 1000 евро"
    return "От 1000 евро"


df["Price_group"] = df["Price_euros"].apply(get_price_group)
```
